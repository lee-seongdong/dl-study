```python
import numpy as np

np.random.seed(0)

N = 10000
D = 2
xs = np.random.rand(N, D) # 균등분포로부터 더미데이터 생성

mu = np.sum(xs, axis=0) # 표본 갯수 축이 0이므로, axis=0
mu /= N

cov = 0

for n in range(N):
    x = xs[n]
    z = x - mu
    z = z[:, np.newaxis]
    cov += z @ z.T

cov /= N

print('계산된 mu:\n', mu)
print('계산된 cov:\n', cov)
print()
print('numpy mu:\n', np.mean(xs, axis=0)) # [D,]
print('numpy cov:\n', np.cov(xs, rowvar=False)) 
```
