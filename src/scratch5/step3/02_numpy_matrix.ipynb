```python
import numpy as np

A = np.array([[1, 2, 3], [4, 5, 6]])
print(A)
print('--- 전치 ---')
print(A.T)
```

```python
A = np.array([[3, 4], [5, 6]])
d = np.linalg.det(A) # 행렬식 determinant: 정사각 행렬의 특징을 나타내는 지표.
print(A)
print('--- 행렬식 ---')
print(d)
```

```python
A = np.array([[3, 4], [5, 6]])
B = np.linalg.inv(A) # 역행렬 inverse matrix: 원래의 행렬과 곱했을때, 단위행렬이 되는 행렬
print(B)
print('--- 역행렬과의 곱 ---') # (M x m) @ (m X N) = (M x N)
print(A @ B) # 행렬곱 (np.dot(a, b))
print((A @ B).astype(int))
```

```python
# mu: 평균벡터
# cov (covariance): 공분산 행렬

# 다변량 정규분포
def multivariate_normal(x, mu, cov):
    det = np.linalg.det(cov)
    inv = np.linalg.inv(cov)
    D = len(x)
    z = 1 / np.sqrt((2 * np.pi) ** D * det)
    y = z * np.exp((x - mu).T @ inv @ (x - mu) / -2.0)
    return y

x = np.array([[0], [0]])
mu = np.array([[1], [2]])
cov = np.array([[1, 0], [0, 1]])

y = multivariate_normal(x, mu, cov)
print(y)
```
