```python
import numpy as np
import matplotlib.pyplot as plt

x_means = []
N = 10

for _ in range(10000):
    xs = []
    for n in range(N):
        x = np.random.rand()
        xs.append(x)
    mean = np.mean(xs) # 표본 평균
    x_means.append(mean)

# 그래프
plt.hist(x_means, bins='auto', density=True) # 히스토그램
plt.title(f'N={N}')
plt.xlabel('x')
plt.ylabel('확률 밀도')
plt.xlim(-0.05, 1.05) # x축 범위
plt.ylim(0, 5) # y축 범위
plt.show()

```
