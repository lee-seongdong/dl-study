```python
import matplotlib.pyplot as plt
import numpy as np

# 다봉 데이터 분포는 여러개의 정규 분포의 조합으로 표현할 수 있다.
# 여러 정규분포로 구성된 모델을 '가우스 혼합 모델 Gaussian Mixture Model (GMM)' 이라고 한다.

# ====== 학습된 매개변수 ======
mus = np.array([[2.0, 54.50],
                [4.3, 80.0]])
covs = np.array([[[0.07, 0.44],
                  [0.44, 33.7]],
                 [[0.17, 0.94],
                  [0.94, 36.00 ]]])
phis = np.array([0.35, 0.65]) # 각 정규분포가 선택될 확률. 총합은 1
# =========================

def sample():
    z = np.random.choice(2, p=phis) # 정규분포 선택 확률에 따라 선택
    mu, cov = mus[z], covs[z]
    x = np.random.multivariate_normal(mu, cov) # 다변량 정규분포에서 임의의 데이터 샘플링
    return x

N = 500
xs = np.zeros((N, 2))
for i in range(N):
    xs[i] = sample()

plt.scatter(xs[:, 0], xs[:, 1], color='orange', alpha=0.7)
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
