```python
import os
import numpy as np
import matplotlib.pyplot as plt

xs = np.loadtxt('height.txt')

# 데이터셋으로부터 평균과 표준편차를 계산
mu = np.mean(xs)
sigma = np.std(xs) 

# 데이터셋의 특징을 가진 확률 밀도 함수로 새로운 데이터 생성
samples = np.random.normal(mu, sigma, 10000)

plt.hist(xs, bins='auto', density=True, alpha=0.7, label='original')
plt.hist(samples, bins='auto', density=True, alpha=0.7, label='generated')
plt.legend()
plt.xlabel('Height(cm)')
plt.ylabel('확률 밀도')
plt.show()

from scipy.stats import norm

# cdf: 정규분포의 누적 분포 함수
# x 이하의 값이 발생할 확률을 반환

p1 = norm.cdf(160, mu, sigma) 
print('p(x <= 160):', p1) # 160 이하의 확률

p2 = norm.cdf(180, mu, sigma)
print('p(x > 180):', 1 - p2) # 180 초과의 확률
```
